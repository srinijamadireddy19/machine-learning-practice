Customer retention cohort analysis tracks how groups of customers defined by a shared starting point or behavior continue to engage with a product or make repeat purchases over time


What is a Cohort?

A cohort is a group of users who share a common characteristic within a defined time frame. Rather than looking at all users in aggregate (which can mask underlying trends), you slice your user base into distinct buckets to see how different groups behave.
